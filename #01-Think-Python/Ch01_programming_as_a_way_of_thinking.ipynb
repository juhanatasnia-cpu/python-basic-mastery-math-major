Integer Division: An operator // that devides two numbers and rounds down to an integer.

Floating point: a type that represents numbers with fractional parts.

Concatenation: Joining two strings end to end.

type: a catagory of values.

operands:nOne of the values on which an operator operates.

Function: A names sequence of statements that performs some useful operations.

Function call: An expression or Part of an expression that runs a function.

expression: A combination of variables, values and operations.

Bug: An error in program.

 Debugging: The process of finding and correcting errors.

Formal languages: Any of the languages that people has designed for specific purpose. Such as Mathematical ideas or computer programs. all programming languages are formal.

Natural Language: Any of the languages that people speak that evoled naturally.



```python
# EXERCISES 01:
# virtual assistant
print(7^2) # the result is 5. 
# Q:1 what are the bitwise operators on Python?
# Q:2 What is the value of 7 XOR 2?
# Q:3 What is the order of operations in python?
# Q:4 What are the arguments of round function?
# Q:5 how do i round pi off to three decimals?
# Q:6 What is the modulus operator in python?

# ​Q:1 What are the bitwise operators in Python?
# ​Bitwise operators treat numbers as strings of binary digits (0s and 1s) and perform operations bit by bit.

# Python Bitwise Operators:
# &  Binary AND
# |  Binary OR
# ^  Binary XOR (Exclusive OR)
# ~  Binary Ones Complement (NOT)
# << Binary Left Shift
# >> Binary Right Shift

#Q:2 What is the value of 7 XOR 2?
#​To find the value, you convert the integers to binary and compare each bit. The XOR operator returns 1 if the bits are different and 0 if they are the same.

# 7 in binary: 111
# 2 in binary: 010
# ----------------
# Result (XOR): 101  (which is 5 in decimal)

print(7 ^ 2) # Output: 5

#Q:3 What is the order of operations in Python?
#​Python follows the PEMDAS rule. For operators with the same precedence, it evaluates from left to right (except for exponentiation).

# 1. Parentheses: ()
# 2. Exponentiation: **
# 3. Multiplication, Division, Modulo: *, /, //, %
# 4. Addition and Subtraction: +, -
# 5. Bitwise operators (lower precedence than arithmetic)

#Q:4 What are the arguments of the round function?
#​The round() function takes two main arguments: the number you want to round and the number of decimal places.

# round(number, ndigits)
# number: The number to be rounded.
# ndigits (optional): The number of decimal places to round to. 
# Defaults to 0 if not provided.

#Q:5 How do I round pi off to three decimals?
#​You can use the math library to get an accurate value of \pi and then apply the round() function.

import math

result = round(math.pi, 3)
print(result) # Output: 3.142

#Q:6 What is the modulus operator in Python?
#he modulus operator yields the remainder of the division between two numbers.
# here in programming like python and modular arithmetic, the "modulus operator" refers to the remainder left over after division.
# The modulus operator is: %

print(7 % 2) # Output: 1 (because 2 goes into 7 three times with 1 left over)
```
