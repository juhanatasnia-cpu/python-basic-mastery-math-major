4. The Modulo Operator:

Modulo Operator: A useful tool to work with numerical information which divides one number by another number and returns the remainder.THe modulo operator doesnt tell us how many times one number fits into another, it only tells us what the remainder is. When one number is divisible by another number, the remainder in 0, so the modulo operator always returns 0. Even numbers are always divisible by 2 so if the modulo of a number and two is zero the number is even. Otherwise it's odd.

```python
number=input("Enter the number:\n")
number=int(number)
if number % 2==0:
    print(f"\n The number {number} is even ")
else: 
    print(f"\n The number {number} is odd")
```

iii. Using a Flag:

flag:For a program that should run as long only as long as many conditions are true, we can define our variable that determines whether or not the entire program is active. This variable is called flag.acts as a signal to the program.We can write our programs so that they can run while thew flag is set to True and stop running when any of several events sets the value of the flag to False. As a result our overall while Statement needs to check only one condition: Whether the flag is currently True.Then all our other tests (to see if an event has occurred that should set the flag to False ) can be neatly organized in the rest of the program.

```python
prompt= "\nTell me something and i will repeat it back to" #Define a prompt that tells the user their two option
prompt+="\nEnter quit to end the program. " # Variable message to keep track of whatever value the user enters.We define message as an empty string, "" so python has something to check the first time it recheases the while line.
active= True
while active:
    message=input(prompt)
    if message== input(prompt):
        active= False
    else:
        print(message)
```

iv. Using break to Exit a loop:To exit a while loop immediately without running any remaining code in the loop, regardless of the results of any conditional test, use the break statement.The break statement directs the flow of our program, we can use it to control which line of code are executed and which aren't, so the program only executes code that we only want it to, when we want it to.

```python
prompt="\nPlease enter your favourite courses:"
prompt+="\n(Enter 'quit) when you are done."
while True:
    course=input(prompt)
    if course=='quit':
        break
    else:
        print(f"{course.title()} is my favourite course too!")
        
```

v. Using Continue in a loop:Rather than breaking out of a loop entirely executing the rest of its code, we can use the continue statement to return to the beginning of the loop, based on the result of a conditional test.

```python
current_number=0
while current_number < 10:
    current_number +=1
    if current_number % 2 ==0:
        continue
    print(current_number)
```

6. Using a while Loop with ists and dictionaries: Allows us to collect, store and organize lots of input to examine and report on later.

Moving items from One List to Another:

```python
# Start with users that need to be verified.
# and an empty list to hold confirmed users.
unconfirmed_users=['edward','alice','kevin']
confirmed_users=[]
#Verify each user until there are no more unconfimed users.
#Move each verified user into the list of confimed users.
while unconfirmed_users:
    current_user=unconfirmed_users.pop()
    print(f"Verifying user:{current_user.title()}")
#Display all confimed users.
print("\nThe following users have been confimed:")
for confirmed_user in confirmed_users:
    print(confirmed_user.title())
```

ii. Removing All instances of specific values from a list:

```python
predators=['lion','tiger','jaguar','panther','tiger','fox','tiger']
print(predators)
while 'tiger' in predators:
    predators.remove('tiger')
print(predators)
```

iii. Filling a Dictionary with user input:

```python
responses={}
# Set a flag to indicate that polling is active:
polling_active=True
while polling_active:
    #Prompt for the person's name and response.
    name=input('\nwhat is your name?')
    response=input("Which mountain would you like to climb someday? ")
    #Store the response in the dictionary.
    responses[name]=response
    #Find out if anyone else is going to take the poll.
    repeat=input("Would you like to let another person respond? (yes/no)")
    if repeat =='no':
        polling_active=False
        #polling is complete. Show the results.
        print("\n--poll Results--")
        for name,response in responses.items():
            print(f"{name} would you like to climb {response}.")
```
