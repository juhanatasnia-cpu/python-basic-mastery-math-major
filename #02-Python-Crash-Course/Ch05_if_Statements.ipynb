Programming often involves examining a set of conditions and deciding which action to take based on thses conditions. Python's if statement allows us to examine the current state of a program and respond appropriately to that state.

Conditional Text: Allows us to check any condition of interest. All heart of every if statement is an expression that can be evaluated as true or false and is called a Conditional test. Python uses the value True and False to decide whether the code is an if statement should be executed. If a conditional test evaluates to True,Python executes the code following the statement. If the statement evaluates to False, Python ignores the code following the if statement.



```python
Courses=['calculus','linear algebra','statistics','geometry']
for course in Courses:
    if course=='calculus':
        print(course.upper())
    else:
        print(course.lower())
    
```

```python
cars=['toyota','bmw','subaru']
for car in cars:
    if car== 'bmw':
        print(car.upper())
    else:
        print(car.title())
```

```python
age_0=22
age_1=18
age_0>=21 or age_1>=21
# The test for age_0 passes, the overall expression evaluates to true.
```

```python
age_0=18
age_0>=21 or age_1>=21
# Both tests now fail and the overall expression evaluates to False.
```

9. **If Statements**:

Several kinds of if statements exist and which is to use depends on the number of conditions we need to text.

i. Simple if Statement:

The simpliest kind of if statement has one test and one action. We can put any conditional test in the first line and just about any action in the indented block following the test. IF the conditional test evaluates to True, Python executes the code following the if statement. If the test evaluates to false, Python ignores the code following the if statement.


```python
age=19
if age>=18:
    print('You are old enough to vote!')
    print('Have you registered to vote yet?')
```

ii. If-else Statements:

There will be times, we'll want to take one action when a conditional test passes and a diferent action in all other cases. Python's if-else statement makes this possible. It allows us to define an action or set of actions that are executed when the conditional test fails.If the conditional test passes, the first block of indented print() calls is executed. If the test evaluates to false, the else block is executed.



```python
age=17
if age>=18:
    print('You are old enough to vote!')
    print('Have you registered to vote yet?')
else:
    print("sorry, you are too young to vote.")
    print("Please register to vote as soon as you turn 18!")
```

iii. if-elif-else Chain:

Often, we will need to test more than two possible situation and to evaluate these we need to use if-elif-else chain. It runs each conditional test in order until one passes. When a test passes, the code following the test is executed and python skips the rest of the tests

```python
age=12
if age<4:
    print("Your Admission cost is $0.")
elif age<18:
    print("Your Admission cost is $25.")
else:
    print("Your Admisson cost is $40.")
```

```python
age=22
if age<4:
    price=0
elif age<18:
    price=25
else:
    price=40
print(f"Your Admission price is ${price}")
```

iv. Using Multiple elif blocks:

We can use as many elif blocks in our code as we like.


```python
age=40
if age<8:
    price=0
elif age<18:
    price=20
elif age>65:
    price=25
else:
    price=40
print(f"Your admission cost is ${price}")
```

10. Testing Multiple conditions:

The if-elif-else is only appropriate to use when we only need one test to pass. As soon as Python finds one test that passes, it skips the rest of the tests.It allows us to test for one specific condition. However, sometimes it's important to check all conditions of interest. In this case we should use a serier of if statements with no elif or else blocks. It makes sense when more than one condition could be true and we want to act on every conditions which is true.

```python
requested_toppings=['mushrooms','extra cheese']
if 'mushrooms' in requested_toppings:
    print("adding mushrooms.")
if 'pepperoni' in requested_toppings:
    print("adding pepperoni.")
if 'extra cheese' in requested_toppings:
    print("adding extra cheese")
print("\nFinished making your pizza")
```

NOTE: If we want only one block of code to run, use an if-elif-else chain. If more than one block of code needs to run, use a series of independent if statements.


11. Using if Statements with Lists:

i. Checking for special items:

```python
requested_toppings=['mashrooms','green peppers', 'extra cheese']
for requested_topping in requested_toppings:
    print(f"adding {requested_topping}.")
print("\nFinished making your pizza!")
```

```python
# If the pizzarie runs out of green peppers. An if statement inside the for loop can handle the situation.
for requested_topping in requested_toppings:
    if requested_topping=='green peppers':
        print("sorry, we are out og green peppers now")
    else:
        print(f"Adding {requested_topping}.")

print("\nFinished making your pizza!")
```

ii. Checking That a list is Not empty:


```python
requested_toppings=[]
if requested_toppings:
   for requested_topping in requested_toppings:
       print(f"adding {requested_topping}.")
       print("\nFinished making your pizza!")
else:
    print("Are you sure you want a plain pizza?")
```

```python
available_toppings=['mashrooms','olives','green peppers','pepperoni','pineapple','extra cheese']
requested_toppings=['mashrooms','french fries','extra cheese']
for requested_topping in requested_toppings:
    if requested_topping in available_toppings:
        print(f"adding {requested_topping}")
    else:
        print(f"sorry, we dont have {requested_topping}")

print("\nfinished making your pizza")
```
