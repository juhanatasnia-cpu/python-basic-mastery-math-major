**How to loop through an entire list**: In case of programming to run same tast with each item through all entires in a list. When we want to do the same action with every item in a list. We use for loop. It's repetative for each item in the list.


```python
Students= ['laura','mark','alice']
for student in Students:
    print(student)
```

2.Doing More Work Within a for loop:

We can write as many lines of code and can do anything with each item.

```python
print(Students)
for student in Students:
    print(f"You did well in exam {student.title()}")
    print(f"Welcome to new the new class")
```

3. Doing sometging after a for loop:

Any lines of code after the loop that are not intendented are executed once without ewpetition.


```python
print(Students)
for student in Students:
    print(f"Congratulation {student}.You got A plus in Math")
    print(f"{student.title()}. You did very good\n")
print('work hard, everyone')
```

NOTE: Python uses indentation to determine how a line or group of lines is related to the rest of the program. Makes code very easy to read. It uses whitespace to force us to write neatly formatted code with a clear visual structure.

7. Working with part of a list:


```python
# Slicing a list: To make slice, we specify the index of the first and last elements we want to work with.
Employees=['alice','martina','charles','ava','nolan']
print(Employees[:2]) # returns elements 0,1
print(Employees[3:])
print(Employees[1:3]) # starting from index 1 to 2
print(Employees[0:2]) 

```

```python
# looping through a slice: We can use a slice in a for loop if we want to loop through a subset of the elements in a list.
workers=['nick','erik','matthus','devid']
print("here are the first three workers:")
for worker in workers[:3]:
    print(worker.title())
```

```python
# Copying a list: Ofthen we will want to start an existing list and make an entirely new list based on the first one to copy a list. We can make a slice that includes the entire original list by ommitting the first index and the second index([:]). This tells python to make a slice that starts at the first item and ends with the last item, producing a copy of the entire list.
my_foods=['pizza','burger','falafel']
friend_foods=my_foods[:]
print("My favourite foods are:")
print(my_foods)
print("\n My friend's favourite foods are:")
print(friend_foods)
```

```python
#to prove each lists are separate:
my_foods.append('ice cream')
friend_foods.append('carrot cake')
print(my_foods)
print(friend_foods)
```

8. Tuples:

Lists work well for storing collections of items that can change throughtout the life of a program. The ability to modify lists is particularly important. Sometimes we want to create a list of items that cannot change. Tuples allow us to do just that. Python refers to values thatt cannot change as immutable and an immutable list is called a tuple.

Tuples just like a list. We use parenthesis (). Once we define a tuple we can access individual elements by using item's index.


```python
dimension=(200, 50) # a rectangle that should always be a certain size. Ensuring it's size never changes.
print(dimension[0])
print(dimension[1])
```

```python
dimensions[0]=250 # when we try to change value of a tuple, we get error.
```

```python
# Looping through all values in a tuple: We can loop over all the values in a tuple using a for loop.
dimensions=(200,50)
for dimension in dimensions:  #Return all elements in the tuple.
    print(dimension)
```

```python
# Writting Over a Tuple; WE can't modify a tuple, we can assign a value to a variable that represents a tuple, if we want to change the dimensions of the rectangle we need to redefine the entire tuple.
dimension=(200,50)
print("Original dimensions:")
for dimension in dimensions:
    print(dimension)

dimensions=(400,100)
print("\nModified dimensions")
for dimension in dimensions:
    print(dimension)
```
