4. Using Variables in Strings:

```python
first_name="Albert"
last_name="Einstein"
full_name=f"{first_name} {last_name}"
print(f"the famouse scientist {full_name} got the noble prize in '1921'")
message=f"{full_name} got noble prize in physics for his groundbreaking discovery for the 'general theory of relativity'"
print(message)
```

4. Adding whitespaces to string with tabs or newlines:

 whitespace: Non printing characters. Such as tabs, spaces and end of line symbols. Organise output so it's easier to read.
 

```python
print("Physics")

print('\tPhysics\n') # \t to add a tab in text.

print("Departments:\n Physics\nMathematics\nBiotechnology\n") #To add a newline using character combination \n

print("Departments:\n\tPhysics\n\tMathematics\n\tBiotechnology") #Combine tabs and newlines in a single string.
```

5. Stripping Whitespaces (Removing extra spaces, manipulating strings):

We use strip() method to remove the extra whitespaces permanently.


```python
favourite_course=' calculus '
favourite_course
```

```python
print(favourite_course.strip())
print(favourite_course.rstrip())
print(favourite_course.lstrip())
```
