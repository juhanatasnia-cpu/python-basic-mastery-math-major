NOTE: 

List: A list is a collection of items. lists allow us to store sets of information in one place. we can  put anything inside a list.

[] square brackets are used in Python to indicate a list and individual elements are separated by commas.

```python
Subjects=['English','Math','Physics','chemistry']
print(Subjects)
```

1. Accerssing elements in a list

Lists are ordered collection. so we can acess any element in a list by telling Python the position or index of the desired item.

```python
# Write the name of the list followed by the index of the item  enclosed in square brackets.
print(Subjects[0])
print(Subjects[1])
print(Subjects[2])
print(Subjects[3])
```

2. Index position starts at 0, not1

```python
print(Subjects[-1])
print(Subjects[-2])
print(Subjects[0])
```

3. Using individual Values from a list

```python
message=f"My favourite subject is {Subjects[1].title()}" #WE can use individual values from a list just as we would with any other variable.
print(message)
```

4. Modifying, Adding and Removing Elements


Modifying elements in a list:

```python
# Modifying elements in a list

print(Subjects)
Subjects[0]='Zoology'
print(Subjects)
```

```python
print(Subjects) # permanently replaced
Subjects[2]='Botany'
print(Subjects)
```

 Adding elements in a List


```python
# Appending elements to the end of a list:
print(Subjects)
Subjects.append('Histry')
print(Subjects)
```

```python
Subjects=[]
Subjects.append('Math')
Subjects.append('Physics')
Subjects.append('Chemistry')
print(Subjects)
```

```python
# inserting elements into a list: To add a new element at any position in a list we can use insert()method by specifying the index of the new element and the value.
print(Subjects)
Subjects.insert(0,'Microbiology')
print(Subjects)
```

Removing elements from a list

```python
# Removing an item using the del statement: If we know the position of the item we want to remove.
Subjects=['Math','Biochemistry','finance']
print(Subjects)
del Subjects[0]
print(Subjects)
del Subjects[1]
print(Subjects)
```

```python
# Removing an item using the pop() Method:Removes the last item in a list but it lets us work with the item after removing it.
Subjects=["Math",'Histry','Finance']
popped_Subjects= Subjects.pop()
print(Subjects)
print(popped_Subjects)
Favourite_Subject= Subjects.pop()
print(f"My favourite Subjects is {Favourite_Subject.title()}")
```

```python
# popping items from any position in a list: Including the index of the item we want to remove in parenthesis
Subjects=['Math','Physics','Economics','CHemistry']
favourite_Subjects=Subjects.pop(0)
print(f"My favourite subject is {favourite_Subjects}")
print(Subjects)
```

```python
#Removing an item by value: When we only know the item we want to remove.

print(Subjects)
Not_interesting='Economics'
Subjects.remove(Not_interesting)
print(Subjects)
print(f"\nI am not interested in {Not_interesting}")
```

5. Organizing a List


```python
# Sorting a list permanently with the sort() method: Changes the order permanently in alphabatical or reverse alphabatical order.
#alphabatically:
Animals=['Lion','Koala','Jaguar','Gorila']
Animals.sort()
print(Animals)

# Reverse Alphabatically
Animals.sort(reverse=True)
print(Animals)
```

```python
# Sorting a list temporaily with the sorted() function: To maintain the original order of a list but present it in a sorted order. Doesn't affect the actual order.
print("Here is the original list:")
print(Animals)
print('\nHere is the sorted list:')
print(sorted(Animals))
print("\nHere is the original list again:")
print(Animals)
```

```python
#Printing a list in Reverse order:To reverse the original order of a list. Changes the order of a list permanently.
print(Animals)
Animals.reverse()
print(Animals)
```

```python
print(Animals)
len(Animals)
# Counts the elements inside the list
```
