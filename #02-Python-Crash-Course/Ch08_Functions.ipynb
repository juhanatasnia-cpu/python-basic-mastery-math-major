1.Function:Functions  are named blocks of code designed to do one specific job. When we want to perform specific task that we have defined in a function, we call the function responsible for it.

Simpliest structure of a function:

```python
def greet_user():
    """Display a simple greeting"""
    print("hellow!")
greet_user()
```

Here the name of the function is greet_user and it needs no information to do its job so the parenthesis are empty. Even so parenthesis are required. 
def greet_user():#Inform python that we are defining a function. This is the function definition, which tells python the name of function and if applicable what kind of information the function needs to do its job.The parenthesis hold that information.

Any indented lines that follows def greet_user(): make up the body of the function. The text on the second line is called docstring, which describes what the function does. When python generates documentation for the function in our programs, it looks for a string immediately after the function's definition. These strings are usually enclosed in triple quotes, which lets us write multiple lines.

When we want to use this function, we have to call it. A function call tells Python to execute the code in the function. To call a function we write the name of the function followd by any necessary information in parenthesis.Because no information is needed here.

3. Argument and Parameters:

Argument: A argument is a piece of information that's passed from a function call to a function. When we call a function we want the function to work within parentheses.

Passing Argument:
 positional Argument: Need to be in the same order the parameters were written.

 Keyword Argument: Where each argument consists of a variable name and a value and list and dictionaries of values.

i. Positional Arguments: When we call a function, Python needs to match each argument in the function call with a parameter in the function definition. The simpliest way to do this is based on the order of the arguments provided. Values matched up this way are called positional arguments.


```python
def describe_pet(animal_type,pet_name):
    """Display information about a pet"""
    print(f"\nI have a {animal_type}")
    print(f"My {animal_type}'s name is {pet_name.title()}")
describe_pet('cat','tom')
```

```python
#Multiple Function call:We can call a function as many times as needed.
def describe_pet(animal_type,pet_name):
    """Display information about a pet"""
    print(f"\nI have a {animal_type}")
    print(f"My {animal_type}'s name is {pet_name.title()}")
describe_pet('cat','tom')
describe_pet('dog','Bull')

# 
```

ii. Keyword Argument:A name value pair that we pass to a function. We directly associate the name and the value within argument, so when we pass the argument to the function, there's no confusion. Keyword arguments free us from having to worry about correctly ordering our arguments in the function call and they clarify the roll of each value in the function call.

```python
def describe_pet(animal_type, pet_name):
    """Display information about a pet"""
    print(f"\nI have a {animal_type}")
    print(f"My {animal_type}'s name is {pet_name.title()}")
describe_pet(animal_type='cat',pet_name='Tom')
```

4. Default Values: 

When writing a function, we can define a default value for each parameter. If an argument for a parameter is provided in the function call, Python uses the argument value.If not, it uses the parameter's default value.So when we define a default value for a parameter, we can exclude the one responding argument we would usually write in the function call. Using default values can simplify our function calls and clarify the ways our function are typically used.

```python
def describe_pet(pet_name, animal_type='dog'):
    """Display information about a pet"""
    print(f"\nI have a {animal_type}")
    print(f"My {animal_type}'s name is {pet_name.title()}")
describe_pet(pet_name='Bull')
```

7. Return Value:

A function doesn't always have to display its output directly. Instead it can process some data and then return a value or set of values. The value the function returns is called a return value. The return statement that takes a value from inside a function and sends it backs to the line that called the function. Return values allow us to move of our program's grunt work into functions,which can simplify the body of our program.

i. Returning a Simple Value:


```python
def get_formatted_name(first_name, last_name):
    """Retuen a full name, neatly formatted"""
    full_name=f"{first_name} {last_name}"
    return full_name.title()
Professor=get_formatted_name('richard','feynman')
print(Professor)
```

ii. Making an Argument Optional:

```python
def get_formatted_name(first_name, middle_name, last_name):
    """Retuen a full name, neatly formatted"""
    full_name=f"{first_name} {middle_name} {last_name}"
    return full_name.title()
actor=get_formatted_name('robert','downey','junior')
print(actor)
```

```python
def get_formatted_name(first_name, last_name, middle_name=''):
    """Retuen a full name, neatly formatted"""
    if  middle_name:
        full_name=f"{first_name} {middle_name} {last_name}"
    else:
        full_name=f"{first_name} {last_name}"
        return full_name.title()
Professor=get_formatted_name('richard','feynman')
print(Professor)
```

iii. Returning a Dictionary:A function can retuen any kind of information we need it to, including more complicated data structures like lists and dictionaries.


```python
def build_person(first_name,last_name):
    """Return a dictionary of information about a person"""
    person={'first':first_name,'last':last_name}
    return person
actor=build_person('Chris','evan')
print(actor)
```

```python
def build_person(first_name,last_name, age=None):
    """Return a dictionary of information about a person"""
    person={'first':first_name,'last':last_name}
    if age:
        person['age']=age
    return person
actor=build_person('Chris','evan',age=38)
print(actor)
```

iv. Using a Function using a while loop:


```python
def get_formatted_name(first_name, last_name):
    """Retuen a full name, neatly formatted"""
    full_name=f"{first_name} {last_name}"
    return full_name.title()
#This is an infinite loop!
while True:
    print("\nPlease tell me your name:")
    f_name=input("First name: ")
    l_name=input("last_name: ")
    formatted_name=get_formatted_name(f_name,l_name)
    print(f"\nHello, {formatted_name}!")

```

```python
def get_formatted_name(first_name, last_name):
    """Retuen a full name, neatly formatted"""
    full_name=f"{first_name} {last_name}"
    return full_name.title()

while True:
    print("\nPlease tell me your name:")
    print("(enter 'q' at any time to quit)")
    f_name=input("First name: ")
    if f_name=='q':
        break
    l_name=input("last_name: ")
    if l_name=='q':
        break
    formatted_name=get_formatted_name(f_name,l_name)
    print(f"\nHello, {formatted_name}!")

```
