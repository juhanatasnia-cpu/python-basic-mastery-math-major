**Dictionary**:A dictionary in Python is a collection of key-value pairs. Each key is connected to a value and we can use a key to access the value associated with that key.A key value can be a number, a string, a list or even another dictionary in fact we can use any object that we can create in python as a value in a dictionary. In python, Dictionaries are wrapped in braces({}) with a series of key-value pairs inside the braces.A key-value pair is a set of values associated with that key. every key is connected to its value by a colon and individual key value pairs are separated by commas. We can store as many key-value pairs as we want in a dictionary even Dictionaries can store limitless amount of information.

1. A simple dictionary:

```python
alien_0= {'color':'green','points':5} #Storing information inside braces.Syntax: 'key':'value'.
print(alien_0['points']) #Accerssing and displaying the informations separately. syntax: print(Dictionary name.[key]),key='points'
print(alien_0['color'])
```

2. Working with Dictionary:

i. Accessing Values in a Dictionary:To get the value associated with a key, we need to give the name of the dictionary and then place the key inside a set of square brackets.


```python
print(alien_0['color'])
new_points=alien_0['points'] #Once the dictionary has been defined. We pull the value associated with the key 'points' from dictionary. This value in then assigned to the variable new_points
print(f"You have earned {new_points} points")
```

ii. Adding New key-value Pairs:Dictionaries are dynamic structures and we can add new key-value pairs to a dictionary at any time. To add a new key-value pairs, we would give the name of the dictionary followed by the new key in square brackets, along with the new value.


```python
print(alien_0) #Main dictionary.
alien_0['x_position']=0  # new pieces of information. The alien's x and y coordinates which will help us display the alien at a particular position on the screen.
alien_0['y_position']=25 #Syntax: Dictionary_name[key]=value. When the key and values are strings,we need to write thwm within quotes ''.
print(alien_0)
```

Dictionaries retain the order in which they were defined.When we print a dictionary or loop through its elements, we will see the elements are in the same order they were added to the dictionaey.

iii. Starting with an empty dictionary:


```python
alien_0={} #Defining a dictionary with empty set of braces.
alien_0['color']='green'
alien_0['points']=5
print(alien_0)
```

iv. Modifying Values in a Dictionary: To modify a value in a dictionary , give the name of the dictionary with the key in square brackets and then the new value we want associated with the key.

```python
alien_0={'color': 'green'}
print(f"The alien is {alien_0['color']}.")
alien_0['color']='yellow'
print(f"The alien is now {alien_0['color']}.")
```

```python
alien_0={'x_position':0, 'y_position':25, 'speed':'medium'}
print(f"Original position:{alien_0['x_position']}")

# Move the alien to the right# Determine how far to move the alien based on its current speed.
if alien_0['speed']=='slow':
   x_increment=1
elif alien_0['speed']=='medium':
   x_increment=2
else:
   # This must be a fast alien.
   x_increment=3
# the new position is the old position plus the increment.
alien_0['x_position']=alien_0['x_position']+x_increment

print(f"New position:{alien_0['x_position']}")
```

v. Removing key-value pairs: When we no longer need a piece of information that's stored in a dictionary, we can use the del statement to completely remove a key-value pair. add del need is the name of the dictionary and the key that we want to remove. 

```python
alien_0={'color':'green','points':5}
print(alien_0)

del alien_0['points'] #Syntax: del dictionary_name[key]
print(alien_0)
```

vi. A dictionary of similar objects: We can use a dictionary to store one kind of information about many objects.


```python
favourite_languages={
    'jen':'python',
    'sarah':'C',
    'Eric':'javascript',
    'phil':'rust',
}
language=favourite_languages['sarah'].title()
print(f"Sarah's favourite language is {language}.")
```

3. Looping through a dictionary:
We can loop through all of a dictionary's key-value pairs. through its keys or through its values.

i. looping through all key-value Pairs:

```python
user_0={
    'username':'efermi',
    'first':'enrico',
    'last':'fermi'
}
for key, value in user_0.items(): # Create names for two variables to print. key and value.The second half of the for statement includes the name of the dictionary followed by the method items() which returns a sequence of key value pairs. 
    print(f"\nKey:{key}")         # The for loop then assigns each of thses pairs to the two variables provided.
    print(f"value:{value}")       # Syntax: for key, value in dictionary_name.items():

```

```python
for name, language in favourite_languages.items():
    print(f"{name.title()}'s favourite language is {language.title()}")
```

ii. Lopping through all key words in a dictionary: The keys() method is useful when we don't need to work with all of the values in a dictionary.

```python
for name in favourite_languages.keys():  # Syntax: for key(here key means name) in dictionary_name.keys():/ for key in dct_name:
    print(name.title())
```

```python
friends=['phil','sarah']
for name in favourite_languages.keys():
    print(f"Hi {name.title()}")

    if name in friends:
        language=favourite_languages[name].title()
        print(f"\t{name.title()},i see you love {language}")
#First, we make a list of friends that we want to print a message to. Inside the loop, we print each person’s name. Then we check whether the name we’re working with is in the list friends . If it is, we determine the person’s favorite language using the name of the dictionary and the current value of name as the key . We then print a special greeting, including a reference to their language of choice.
```

```python
# We can also use the keys()  method to find out if a particular person was polled.
if 'erin' not in favourite_languages.keys():
    print("erin, please take our poll")
```

iii. Looping through a dictionary's keys in a particuilar order: Looping through a dictionary returns the items in the same order they were inserted. Sometimes we will want to loop through a dictionary in a different order.By using the sorted() function to get a copy of the keys in order, sorting the keys as they are returned in the for loop.

```python
for name in sorted(favourite_languages.keys()):
    print(f"{name.title()}, thank you for taking the poll")
```

iv. Looping through all values in a dictionary: We can use the values() method to return a sequence of values without any keys.


```python
print("The following language have been mentioned:")
for language in favourite_languages.values(): # Before we returned keys only. now we only returned values.
    print(language.title()) # Syntax: for value(here the values mean languages) in dictionary_name.values():
```

```python
print("The following language have been mentioned:")
for language in set(favourite_languages.values()): # To see each language chosen without repetition, we use a set. A set is a collection in which each item must be unique.
    print(language.title())
    
```

4. Nesting:Multiple dictionaries in a list, or a list of items as a value in a dictionary. This is called nesting.We can nest dictionaries inside a list, a list of items inside a dictionary, or even a dictionary inside another dictionary.

i. A list of Dictionaries:




```python
alien_0={'color':'green','points':5}
alien_1={'color':'yellow','points':10}
alien_2={'color':'red','points':15}
aliens=[alien_0,alien_1,alien_2]
for alien in aliens:
    print(alien)
```

```python
# Making an empty list for storing aliens.
aliens=[]
# Make 30 green aliens.
for alien_number in range(30):
    new_alien={'color':'green','points':5,'speed':'slow'}
    aliens.append(new_alien)
#Show the first 5
for alien in aliens[:5]:
    print(alien)
print("...")
#Show how many aliens have been created.
print(f"Total number of aliens:{len(aliens)}")

```

ii. A list in a dictionary:Rather than putting a dictionary inside a list, it's sometimes useful to put a list inside a dictionary.

```python
#Store information about a pizza being ordered.
pizza={
    'crust':'thick',
    'toppings':['mushrooms','extra cheese']
}
#Summarize the order
print(f"you ordered a {pizza['crust']}-crust pizza"
      "with the following toppings:")
for topping in pizza['toppings']:
    print(f"\n{topping}")
    
```

```python
favourite_languages={
    'jen':['python','c'],
    'sophia':['C','rust'],
    'phil':['go'],
    'edward':['python','javascript'],
}
for name, languages in favourite_languages.items():
    print(f"\n{name.title()}'s favourite languages are:")
    for language in languages:
        print(f"\t{language.title()}")
```

iii. A Dictionary in a Dictionary:

```python
users={
    'einstein':{
        'first':'albert',
        'last':'eintein',
        'location':'princeton',
    },
    'mcurie':{
        'first':'marie',
        'last':'curie',
        'location':'paris',
    },
}
for username, user_info in users.items():
    print(f"\nUsername:{username}")
    full_name=f'{user_info['first']} {user_info['last']}'
    location=user_info['location']
    print(f"\tFull name:{full_name.title()}")
    print(f"\tLocation:{location.title()}")
```
